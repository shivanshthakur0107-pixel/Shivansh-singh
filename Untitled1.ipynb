# SET operation
## 1. union
## 2. INteraction
## 3. Difference
## 4. Complement
## 5. Subset

```python
# now create two set
python_students={"Rahul","Manish","Anil","Chirag"}
java_students={"Manish","Lalit","Anil","Rajith","Sanjiv"}


# union operation
result=python_students.union(java_students)
print("Union Operation:",result)

#intersection
result_inter=python_students.intersection(java_students)
print("Result of intersection operation:",result_inter)


#difference
res_diff=python_students.difference(java_students)
print("Result of difference:",res_diff)
```
