```python
import numpy as np
import pandas as pd
```

```python
# create your own dataset


data={'Student_ID':[101,102,103,104,105,106,107,108,109,110,111,112],
    'Name':['Rahul kumar','Amit sharma','Priya singh','Neha verma','Ravi','ANjali gupta','VIkas singh','Pooja shrama','Karan mehta','Sneha kumari','Arjun singh','Riya verma'],

    'Gender':['Male','Male','Female','Female','Male','Female','Male','Female','Male','Female','Male','Female'],
    'Department':['CSE','CSE','ECE','CSE','ECE','CSE','ME','CSE','ECE','ME','CSE','ECE',],
    'Age':[20,21,np.nan,22,21,20,23,np.nan,22,21,20,22],
    'Attendence':[85,72,np.nan,75,88,65,92,np.nan,76,64,95,81],
    'Marks':[85,72,np.nan,91,68,88,55,np.nan,76,64,95,81],
    'Email':['rahul@gmail.com','amit@gmail.com','priya@gmail.com','neha@gmail.com','None','anjali@gmail.com','vikas@gmail.com','pooja@gmail.com','None','None','None','None']
}
df=pd.DataFrame(data)
display(df)

```

```python
print('Shape:',df.shape)
print('Columns:',df.columns.tolist)
print('Data Types:',df.dtypes)
display(df.head())
display(df.tail())

```

```python
# Detect Missing values
print(df.isnull().sum())
```

```python
#Handle the missing values
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Attendence']=df['Attendence'].fillna(df['Attendence'].median())
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df['Email']=df['Email'].fillna('Not Provided')
print(df.isnull().sum())

```

```python
#data Transformation
df['Result']=df['Marks'].apply(lambda x:'PAss' if x>=40 else 'Fail')


def assign_grade(marks):
  if marks>=80:return 'A'
  elif marks>=60:return 'B'
  elif marks>=60:return 'C'
  elif marks>=50:return 'D'
  else:return 'F'
df['Grade']=df['Marks'].apply(assign_grade)
display(df)
```
